```python
import pandas as pd
import numpy as np

# 1. Connect IBM's variables to our working variables
df_train = df_1.copy()
df_test = df_2.copy() 

# 2. Map specific attack types ONLY on the train set (test set has no labels)
attack_mapping = {
    'normal': 'Normal',
    'neptune': 'DoS', 'smurf': 'DoS', 'pod': 'DoS', 'teardrop': 'DoS', 'land': 'DoS', 'back': 'DoS', 'apache2': 'DoS', 'udpstorm': 'DoS', 'processtable': 'DoS', 'mailbomb': 'DoS',
    'satan': 'Probe', 'ipsweep': 'Probe', 'portsweep': 'Probe', 'nmap': 'Probe', 'mscan': 'Probe', 'saint': 'Probe',
    'guess_passwd': 'R2L', 'ftp_write': 'R2L', 'imap': 'R2L', 'phf': 'R2L', 'multihop': 'R2L', 'warezmaster': 'R2L', 'warezclient': 'R2L', 'spy': 'R2L', 'xlock': 'R2L', 'xsnoop': 'R2L', 'snmpguess': 'R2L', 'snmpgetattack': 'R2L', 'httptunnel': 'R2L', 'sendmail': 'R2L', 'named': 'R2L',
    'buffer_overflow': 'U2R', 'loadmodule': 'U2R', 'perl': 'U2R', 'rootkit': 'U2R', 'xterm': 'U2R', 'ps': 'U2R', 'sqlattack': 'U2R'
}

# The Kaggle dataset usually names the target column 'class'
target_col = 'class'

# Apply mapping and drop rows where the attack type wasn't recognized
df_train['Category'] = df_train[target_col].map(attack_mapping)
df_train = df_train.dropna(subset=['Category'])

# Save the answers (targets) for training
y_train_full = df_train['Category']

# Drop the labels from train so it matches the test set structurally
df_train = df_train.drop(columns=[target_col, 'Category'])

# 3. Add markers and combine them safely for text encoding
df_train['is_train'] = 1
df_test['is_train'] = 0

df_combined = pd.concat([df_train, df_test], ignore_index=True)
print("Data safely combined! Shape:", df_combined.shape)
```

```python
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

# 1. Handle categorical features for both sets simultaneously
categorical_cols = ['protocol_type', 'service', 'flag']

for col in categorical_cols:
    le = LabelEncoder()
    df_combined[col] = le.fit_transform(df_combined[col].astype(str))

# 2. Split them back up using our marker
train_features = df_combined[df_combined['is_train'] == 1].drop(columns=['is_train'])
test_features_final = df_combined[df_combined['is_train'] == 0].drop(columns=['is_train'])

# 3. Create a Validation split from the Train data (so we can check accuracy!)
X_train, X_val, y_train, y_val = train_test_split(train_features, y_train_full, test_size=0.2, random_state=42)

# 4. Scale the numerical features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)

print("Preprocessing complete!")
print(f"Training shape: {X_train_scaled.shape}")
print(f"Validation shape: {X_val_scaled.shape}")
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score

print("Training Random Forest Classifier... (This may take a minute)")
rf_model = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
rf_model.fit(X_train_scaled, y_train)

# Evaluate against our Validation set
y_pred = rf_model.predict(X_val_scaled)

print("\n--- Model Evaluation ---")
print(f"Accuracy: {accuracy_score(y_val, y_pred) * 100:.2f}%\n")
print("Classification Report:")
print(classification_report(y_val, y_pred))
```
